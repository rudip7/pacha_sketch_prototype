```python
import re
from collections import Counter

def extract_authors(text):
    authors = []
    # Each reference starts with [number]
    entries = re.split(r'\[\d+\]', text)
    for entry in entries:
        entry = entry.strip()
        if not entry:
            continue
        # Extract everything before the first 4-digit year
        match = re.match(r'^(.*?)(\d{4})', entry, re.DOTALL)
        if not match:
            continue
        author_part = match.group(1).strip()
        # Replace " and " with a comma for uniform splitting
        author_part = author_part.replace(" and ", ", ")
        # Split by comma
        raw_authors = [a.strip() for a in author_part.split(",") if a.strip()]
        authors.extend(raw_authors)
    return authors

def rank_authors(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        text = f.read()
    authors = extract_authors(text)
    counter = Counter(authors)
    for author, count in counter.most_common():
        print(f"{author}: {count}")

if __name__ == "__main__":
    rank_authors("references.txt")
```
