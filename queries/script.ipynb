```python
import json
import os

def transform_queries(obj):
    if isinstance(obj, list):
        new_list = []
        for item in obj:
            # If it's a list with exactly one integer, turn it into a string
            if isinstance(item, list) and len(item) == 1 and isinstance(item[0], int):
                new_list.append([str(item[0])])
            else:
                new_list.append(transform_queries(item))
        return new_list
    elif isinstance(obj, dict):
        return {k: transform_queries(v) for k, v in obj.items()}
    else:
        return obj

def process_directory(directory):
    for root, _, files in os.walk(directory):  # 🔹 walks into all subfolders
        for filename in files:
            if filename.endswith(".json"):
                filepath = os.path.join(root, filename)
                print(f"Processing {filepath}...")

                with open(filepath, "r") as f:
                    data = json.load(f)

                # Transform
                data = transform_queries(data)
                print(f"Done: {filename}")

                # Overwrite the same file (or change this to a new filename if you want backups)
                with open(filepath, "w") as f:
                    json.dump(data, f)
    
```

```python
input_dir = "census" 
process_directory(input_dir)
```
